# Integrated FP&A Budgeting, Forecasting & Scenario Planning

This notebook demonstrates the decision layer of the synthetic Asteria Consumer Group FP&A portfolio project.

```python
from pathlib import Path
import pandas as pd

DATA = Path('../Data')
monthly = pd.read_csv(DATA / 'monthly_pnl.csv', parse_dates=['Month'])
scenarios = pd.read_csv(DATA / 'scenario_summary.csv')
models = pd.read_csv(DATA / 'forecast_model_comparison.csv')
risk = pd.read_csv(DATA / 'risk_summary.csv')
```

## FY2026 budget versus rolling forecast

```python
fy2026 = monthly[monthly['Month'].dt.year.eq(2026)]
summary = fy2026.groupby('Version')[['RevenueTRY', 'GrossProfitTRY', 'EBITDATRY', 'NetIncomeTRY']].sum()
summary.assign(EBITDAMarginPct=summary['EBITDATRY'] / summary['RevenueTRY'])
```
